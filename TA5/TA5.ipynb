```python
from textblob import TextBlob


def classify_sentiment(sentence):
    """Return the polarity and its positive, neutral, or negative label."""
    polarity = TextBlob(sentence).sentiment.polarity

    if polarity < 0:
        label = "Negative"
    elif polarity == 0:
        label = "Neutral"
    else:
        label = "Positive"

    return polarity, label


def main():
    counts = {"Positive": 0, "Neutral": 0, "Negative": 0}

    for number in range(1, 11):
        sentence = input(f"Enter sentence {number}: ")
        polarity, label = classify_sentiment(sentence)
        counts[label] += 1
        print(f"Polarity: {polarity:.2f} - {label}\n")

    print("Sentiment Summary")
    print(f"Positive: {counts['Positive']}")
    print(f"Neutral:  {counts['Neutral']}")
    print(f"Negative: {counts['Negative']}")


if __name__ == "__main__":
    main()
```

```python
import re

import nltk
from nltk.corpus import opinion_lexicon


nltk.download("opinion_lexicon", quiet=True)
HAPPY_WORDS = set(opinion_lexicon.positive())
SAD_WORDS = set(opinion_lexicon.negative())


def sentence_words(sentence):
    """Return lowercase words and words containing a hyphen or apostrophe."""
    return re.findall(r"[a-z]+(?:[-'][a-z]+)*", sentence.lower())


def number_happy(sentence):
    """Count words that appear in the positive-word dataset."""
    return sum(word in HAPPY_WORDS for word in sentence_words(sentence))


def number_sad(sentence):
    """Count words that appear in the negative-word dataset."""
    return sum(word in SAD_WORDS for word in sentence_words(sentence))


def main():
    sentence = input("Enter a sentence: ")
    happy_count = number_happy(sentence)
    sad_count = number_sad(sentence)

    if happy_count > sad_count:
        tone = "Happy"
    elif sad_count > happy_count:
        tone = "Sad"
    else:
        tone = "Neutral"

    print(f"Happy words: {happy_count}")
    print(f"Sad words: {sad_count}")
    print(f"Tone: {tone}")


if __name__ == "__main__":
    main()
```
